## Problem 1 

A little introductory brain teaser. Which is more probable when rolling 2 six-sided dice: rolling snake eyes (two ones) or rolling sevens (dice sum to seven)? What is the ratio of the probabilities?

The probability of rolling two ones is a product of the individual probabilities, so P(1 and 1)=P(1)P(1)=(1/6)(1/6)=1/36. 

For having the dice sum to seven, there are a number more combinations, being: 
1 and 6;
2 and 5;
3 and 4;
4 and 3;
5 and 2;
6 and 1.
These probabilities add, so the probability of having the dice sum to seven is:
P(1 and 6) + P(2 and 5)+ P(3 and 4)+ P(4 and 3) + P(5 and 2)+ P(6 and 1)
= 6(1/36)=1/6


So the ratio of rolling sevens to rolling snake eyes is (1/6)/(1/36)=6

## Problem 2
Following what we did in class show how to use the convolution operator to determine the probability of the sum of 2 six sided dice. Do both analytically (math & counting) and numerically (computer program). Beware the implicit definition of the values (x-axis on histogram).



In class we used $ P_{A+B}(x) = \int_{-\infty}^{\infty} P_A(z) P_B(x - z) \, dz = P_A(x) * P_B(x) $, but since dice have discrete probability values I can rewrite the integral as $P_{A+B}(x)=\sum _{z=1}^{6}P_{A}(z)P_{B}(x-z)$.
So analytically, 

For sum x=2, $P_{A+B}(2)=\sum _{z=1}^{6}P_{A}(z)P_{B}(2-z)$, and since the dice can't roll a number less than 1, the the only nonzero term is if z=1, so $ P_A(1)P_B(1) = \frac{1}{6} \times \frac{1}{6} = \mathbf{\frac{1}{36}}$.

For sum x=3:
$P_{A+B}(3)=\sum _{z=1}^{6}P_{A}(z)P_{B}(3-z)$, and the only nonzero term is if z=1 or 2, so: $P_A(1)P_B(1)+P_A(2)P_B(1) = \frac{1}{6} \times \frac{1}{6} +  \frac{1}{6} \times \frac{1}{6} = \mathbf{\frac{1}{36}}+ \mathbf{\frac{1}{36}}=\mathbf{\frac{1}{18}} $.

For sum x=7:

$P_{A+B}(7)=P_{A}(1)P_{B}(6)+P_{A}(2)P_{B}(5)+P_{A}(3)P_{B}(4)+P_{A}(4)P_{B}(3)+P_{A}(5)P_{B}(2)+P_{A}(6)P_{B}(1)$
$= 6 ( \mathbf{\frac{1}{36}})$
$=1/6$

For sum x=12:

$P_{A+B}(12)=P_{A}(6)P_{B}(6)=\mathbf{\frac{1}{36}}$

Numerically:

```python
import numpy as np
import matplotlib.pyplot as plt
pdf_die = np.array ([1/6, 1/6, 1/6, 1/6, 1/6, 1/6])
convolution=np.convolve(pdf_die, pdf_die)

print("Sum|  Probability")
for s in range(0, 11):
    print(f"{s}  | {convolution[s]} ({int(round(convolution[s]*36))}/36)")
    sums = np.arange (2,13)
plt.bar(sums, convolution, color='skyblue', edgecolor='black', width=0.8)

plt.xticks(sums)

plt.title('Probability Distribution of the Sum of Two Dice')
plt.xlabel('Dice Sum Value')
plt.ylabel('Probability')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

## Problem 5:

Show that the sum and average of an initially Gaussian distribution is also a Guassian (can be analytic or numerical). How does the standard deviation of the resulting sum or average Gaussian change? This is a hugely important result. Explore what this means for integrating a signal over time. How does this explain error analysis and error propagation you have done in previous classes (classes that assumed Gaussian errors)?

For two Gaussian distributions $f(x) = \frac{1}{\sigma_f \sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x-\mu_f}{\sigma_f}\right)^2}
$ and $g(x) = \frac{1}{\sigma_g \sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x-\mu_g}{\sigma_g}\right)^2}
$ , so summing them gives:
$f(x) + g(x) = \frac{1}{\sigma_g \sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x-\mu_g}{\sigma_g}\right)^2}$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 1000)

def gaussian_pdf(x, mu, sigma):
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

# . parameters for N = 2 identical distributions (mu = 0, sigma = 1)
# Base: mu = 0, sigma = 1
pdf_base = gaussian_pdf(x, mu=0, sigma=1)

# Sum: mu = 0 + 0 = 0, sigma = sqrt(1^2 + 1^2) = sqrt(2)
pdf_sum = gaussian_pdf(x, mu=0, sigma=np.sqrt(2))

# Average: mu = (0 + 0)/2 = 0, sigma = sqrt(1^2 + 1^2)/2 = sqrt(2)/2
pdf_avg = gaussian_pdf(x, mu=0, sigma=np.sqrt(2) / 2)


plt.figure(figsize=(10, 6))

plt.plot(x, pdf_base, label='Base Distribution N(0,1)', color='blue', linewidth=2)
plt.plot(x, pdf_sum, label='Sum of Two Gaussians N(0,2)', color='red', linewidth=2, linestyle='--')
plt.plot(x, pdf_avg, label='Average of Two Gaussians N(0,0.5)', color='green', linewidth=2, linestyle=':')


plt.title('Gaussian Behavior: Base vs. Sum vs. Average (N=2)', fontsize=14, fontweight='bold')
plt.xlabel('Value (x)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)


plt.tight_layout()
plt.show()
```

This is an important result because it means that gaussian inputs are still gaussian even if they are summed or averaged, so they can be easily distringuished from other signals or events.
