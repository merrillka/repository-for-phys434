```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

#this sets the size of the plot to something useful
plt.rcParams["figure.figsize"] = (15,10)

mu = 10.0 #the mean of the gaussian
sigma = 0.5 #the standard deviation of the gaussian
d = stats.norm.rvs(loc=mu, scale=sigma, size=1000000)
```

A Gaussian's probability density $$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}}e^{-\frac{(x-\mu)^2}{2\sigma^2}} $$ has $\mu$ as the mean and the center of the distribution, and $\sigma$ as the standard dev and the width of the distribution. For the new central value, I changed the loc from 5 to 10 for the new $\mu$ , and for the new standard deviation I changed the scale from 0.01 to 0.5 for the new $\sigma$. The curve makes a symmetric fit over the data, and after around  $\mu\pm 3\sigma$ the linear plot is difficult to see and I can't tell whether the tails are also a Gaussian fit.

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d, 50, density=True, label="samples (histogram)")
plt.tick_params(labelsize = 24)
plt.xlim([mu - 5*sigma, mu + 5*sigma])
x = np.linspace(mu - 5*sigma, mu + 5*sigma, 1000)
ax.plot(x,stats.norm.pdf(x,loc = mu, scale = sigma),linewidth = 8,alpha = 0.7, label=r"curve: $f(x)$")
ax.set_xlabel("x")
ax.set_ylabel("probability density (1 / unit of x)")
ax.set_title(rf"samples, Gaussian with $\mu = {mu}$, $\sigma = {sigma}$ (linear scale)")
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
ax.set_xlabel('x')
ax.set_ylabel('probability density')
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([mu - 5*sigma, mu + 5*sigma]) 
x = np.linspace(mu - 5*sigma, mu + 5*sigma, 1000)
ax.plot(x, stats.norm.pdf(x, loc=mu, scale=sigma), linewidth=8, alpha=0.7)
ax.set_title("Histogram log scale")
plt.show()
```

The histogram plot is generated by the fig, ax=plt.subplots (1,1) line, and ax.hist creates the histogram. I updated the plot ranges to [$\mu$-5 $\sigma$, $\mu$+5 $\sigma$ ] . The log scale in the third line converts the y-axis to a logarithmic scale so that we can check the tails of the Gaussian plot  $$ f(x) = \frac{1}{\sqrt{2\pi\sigma^2}}e^{-\frac{(x-\mu)^2}{2\sigma^2}} $$  to check for a Gaussian fit like we discussed in the first lecture.  On the tails of the plot, the log axis stretches the data, so there are less samples and $1/\sqrt{n}$ has a more scattered fit. This Gaussian looks like an upside down parabola on the log plot since $\ln f(x) = - \frac{(x-\mu)^2}{2\sigma^2}+\text{const}$. The stretched out curve still agrees pretty much with the data in the center, at the vertex of the parabola  at x=$\mu$.
