## 2a) What is the pdf of the sum of two identical exponential distributions?
Looking at this
[Link] Wikipedia Page (https://en.wikipedia.org/wiki/Exponential_distribution)
, the section on the sum of two independent exponential random variables shows that the pdf of the two is the same as the convovlution of their individual pdfs. Since the pdf of an exponential distribution is $\lambda e^{-\lambda x}$, their convolution will be $\lambda^2 x e ^{-\lambda x}$ which is the pdf of the Erlang distribution wit shape k=2



## 2b) What is the pdf of the ratio of two zero-mean unity variance normal distributions X1/X2?


For a fixed $X_2 = x_2$ the ratio $z = x_1/x_2$ has $dx_1 = |x_2|\,dz$, so

$$f_Z(z) = \int_{-\infty}^{\infty} |x_2|\, \frac{1}{2\pi} e^{-\frac{x_2^2 (1+z^2)}{2}}\,dx_2 = \frac{1}{\pi}\int_0^\infty x_2\, e^{-\frac{x_2^2 (1+z^2)}{2}}\,dx_2 = \frac{1}{\pi(1+z^2)}.$$
which is the standard Cauchy distribution,

The Cauchy distribution has no defined mean or variance, so averages of Cauchy samples do not converge as the number of samples grows.

## 2c) So far we have looked at 1D probability distributions, but it is possible to have a multi-dimensional vector distribution. A simple first introduction is the 2D Guassian; it looks like a smeared spot. If I measure the amplitude of this vector, what is its pdf? (Hint, the amplitude is always positive.)
  
The joint pdf is $\frac{1}{2\pi}e^{-(x^2+y^2)/2}$. In polar coordinates $x^2 + y^2 = r^2$ and $dx\,dy = r\,dr\,d\theta$, so integrating over the angle $\theta$ from $0$ to $2\pi$ gives

$$f_R(r) = r\, e^{-r^2/2}, \qquad r \ge 0.$$

This is the Rayleigh distribution with $\sigma = 1$ (for general $\sigma$, $f_R(r) = \frac{r}{\sigma^2}e^{-r^2/2\sigma^2}$).
